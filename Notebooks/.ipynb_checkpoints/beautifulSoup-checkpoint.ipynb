### `.contents` and `.children`

A tag's children are available in a list called .contents:

```python
head_tag.contents[:4]
```

```python
len(soup.contents)
```

```python
soup.contents[0].name
```

```python
head_tag.children
```

```python
head_tag = soup.head
for child in head_tag.descendants:
    print(child)
```

```python
title_tag = soup.title
```

```python
# If a tag has only one child, and that child is a NavigableString, we can use .string
title_tag.string
```

```python
# If a tag's only child is another tag, and that tag has a .string, then the parent tag is considered to have the same .string 
print(head_tag.contents)
print(head_tag.string)
```

```python
# If a tag contains more than one thing, .string returns None
print(soup.html.string)
```

# Searching the tree

The two most popular methods are `find()` and `find_all()`.

## Kinds of filters
- A string: pass a string to a search method, and Beautiful Soup will perform a match against that exact string.
- A regular expression: we can use `re.compile()` to pass a regex.
- A list: Beautiful Soup will match against any item in that list.
- A function: a function return `True` or `False`

    

```python
def has_class_but_no_id(tag):
    return tag.has_attr('class') and not tag.has_attr('id')
```

```python
soup.find_all(has_class_but_no_id)
```

```python
soup.a.has_attr('id')
```

```python
def is_the_only_string_within_tag(s):
    return (s == s.parent.string)
```

```python
soup_e.find_all(string=is_the_only_string_within_tag)
```
